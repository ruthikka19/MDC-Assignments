1. Numbers between 2000 and 3200 divisible by 7 and not a multiple of 3

```python
result=[]
for num in range(2000, 3201):
  if num%7==0 and num%3!=0:
    result.append(str(num))
print(",".join(result))
```

2. Convert a List of Characters into a String


```python
char_list=["R","U","T","H","I","K","K","A"]
result_str=""
for char in char_list:
  result_str += char
print(result_str.capitalize())
```

3. Generate a Dictionary of Squares (i, i x i)

```python
n=5
dictionary={}
for i in range(1,n+1):
  dictionary[i]=i*i
print(dictionary)
```
